# 03 — Submission
Jalankan notebook modeling atau CLI training terlebih dahulu. Evaluator meminta integer 1–5 walaupun sample menggunakan teks Grade N. Cell prediksi di bawah akan mengganti submissions/submission_native.csv.

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'scripts').is_dir() else Path.cwd().parent
assert (ROOT / 'scripts').is_dir(), 'Buka notebook dari root proyek atau notebooks/'
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from scripts.predict import generate_submission
MODEL_PATH = ROOT / 'models/native_ensemble.pkl'
assert MODEL_PATH.exists(), 'Latih pipeline terlebih dahulu melalui scripts.train'
submission = generate_submission(model_path=MODEL_PATH,
                                 output=ROOT / 'submissions/submission_native.csv')
submission.head()
```

```python
import pandas as pd
sample = pd.read_csv(ROOT / 'data/sample_submission.csv')
assert submission.columns.tolist() == ['building_id', 'damage_grade']
assert submission.building_id.equals(sample.building_id)
assert submission.building_id.is_unique
assert submission.damage_grade.between(1, 5).all()
assert pd.api.types.is_integer_dtype(submission.damage_grade)
assert not submission.isna().any().any()
print(f'Valid: {len(submission):,} baris. Upload submissions/submission_native.csv ke Kaggle.')
submission.damage_grade.value_counts().sort_index()
```
